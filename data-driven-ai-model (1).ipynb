```python
def predict_next_day(past_temperatures, window=3):
    """Predicts tomorrow's temperature as the average of the last 'window' data"""
    recent = past_temperatures[-window:]
    prediction = sum(recent) / len(recent)
    return round(prediction, 1)

if __name__ == "__main__":
    #last 7 days of recorded temperatures (in celcius)
    past_week = [50, 37,34,46,43,21,34,19]
    actual_tomorrow = 31 #what the temperature actually turned out to be
    predicted = predict_next_day(past_week, window=3)
    error = round(abs(predicted - actual_tomorrow), 1)
    print(f"Predicted: {predicted}C")
    print(f"Actual:    {actual_tomorrow}C")
    print(f"Error:     {error}C")
```

```python
keywords = ["free", "winner", "urgent", "click here", "prize"]
threshold = 1


def predict_spam(text):
    count = 0
    # Count how many keywords appear in the email
    for word in keywords:
        if word in text.lower():
            count += 1

    # Compare count against threshold
    if count >= threshold:
        return "spam"
    else:
        return "not spam"


# Sample dataset: (email text, correct label)
sample_emails = [
    ("this email contains a free surprise for you", "spam"),
    ("urgent request", "spam"),
    ("Can we reschedule our project sync meeting to 3 PM tomorrow?", "not spam"),
    (
        "Congratulations! You are the official winner of our giveaway. Click here to claim.",
        "spam",
    ),
    ("Please review the attached Q3 financial performance report.", "not spam"),
    ("Claim your mystery prize today before time runs out!", "spam"),
]

correct_count = 0

# Test each email and compare prediction to actual label
for text, actual in sample_emails:
    predicted = predict_spam(text)

    if predicted == actual:
        correct_count += 1

    print(f"Text: '{text}'")
    print(f"Predicted: {predicted} | Actual: {actual}\n")

# Calculate and report accuracy
accuracy = (correct_count / len(sample_emails)) * 100
print(f"Accuracy: {accuracy:.2f}%")
```
