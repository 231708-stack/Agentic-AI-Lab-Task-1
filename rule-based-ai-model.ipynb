```python
def weather_advice(temperature_c, humidity_percent):
    advice = []
    
    if temperature_c >= 0 and temperature_c <= 35 :
        advice.append("Extreme heat: Stay hydrated and avoid the sun")
    elif temperature_c <0 and temperature_c>-15 :
        advice.append("Cold Weather: wear a warm jacket")
    elif temperature_c <= -15:
        advice.append("Extreme Cold: Stay Inside")
    elif temperature_c >= 50:
        advice.append("Extreme Hot: Stay Indoors")
        
    if humidity_percent >= 80:
        advice.append("High Humidity: expect possible rain, carry an umbrella")
        
    if not advice:
        advice.append("Conditions look moderate, No special precautions needed.")
        
    return advice

if __name__ == "__main__":
    sample_days = [(-20, 80), (60, 65), (18, 95), (12, 40)]
    for temp, humidity in sample_days:
         result = weather_advice(temp, humidity)
         print(f"Temp={temp}C, Humidity={humidity}% -> {result}")
```

```python
def weather_advice(temperature_c, humidity_percent, wind_speed_kmh, aqi):
    advice = []
    
    # Temperature rules (adjusted slightly from original to fix logical gaps)
    if 35 <= temperature_c < 50:
        advice.append("Extreme heat: Stay hydrated and avoid the sun")
    elif -15 < temperature_c < 0:
        advice.append("Cold Weather: wear a warm jacket")
    elif temperature_c <= -15:
        advice.append("Extreme Cold: Stay Inside")
    elif temperature_c >= 50:
        advice.append("Extreme Hot: Stay Indoors")
        
    # Humidity rule
    if humidity_percent >= 80:
        advice.append("High Humidity: expect possible rain, carry an umbrella")
        
    # New Rule 1: Wind Speed
    if wind_speed_kmh >= 45:
        advice.append("High Wind: Secure loose objects outdoors and drive carefully")
        
    # New Rule 2: Air Quality Index (AQI)
    if aqi >= 150:
        advice.append("Poor Air Quality: Wear a mask and avoid strenuous outdoor activity")
        
    # Default case
    if not advice:
        advice.append("Conditions look moderate, No special precautions needed.")
        
    return advice

if __name__ == "__main__":
    # Test cases formatted as: (Temperature, Humidity, Wind Speed, AQI)
    sample_days = [
        (22, 45, 10, 40),    # Test 1: Moderate conditions
        (-5, 50, 50, 30),    # Test 2: Cold and highly windy
        (38, 85, 15, 160),   # Test 3: Extreme heat, high humidity, poor air quality
        (-20, 30, 5, 20),    # Test 4: Extreme cold
        (52, 20, 60, 95)     # Test 5: Extreme hot and high wind
    ]
    
    print("--- Weather Advisor Test Log ---\n")
    for temp, humidity, wind, aqi in sample_days:
        result = weather_advice(temp, humidity, wind, aqi)
        print("\n")
        # Printing inputs and results clearly as requested in Image_20260914_095016_837.jpg
        print(f"Inputs: Temp={temp}°C | Humidity={humidity}% | Wind={wind}km/h | AQI={aqi}")
        print("Advice:")
        for advice_string in result:
            print(f" - {advice_string} ")
```
